```python
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download("omw-1.4")

data = pd.read_csv(r"D:\Jupyter\NLP Skills\Hometasks\IMDB_Dataset_CLEANED.csv")

print("\nOriginal Dataset:")
print(data.head())

print("\nDataset Shape:")
print(data.shape)

print("\nMissing Values:")
print(data.isnull().sum())

data = data.dropna()

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def clean_review(text):
    text = text.lower()
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    words = text.split()

    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)


data["cleaned_review"] = data["review"].apply(clean_review)

print("\nOriginal Review:")
print(data["review"].iloc[0])

print("\nCleaned Review:")
print(data["cleaned_review"].iloc[0])

data["sentiment"] = data["sentiment"].map({
    "positive": 1,
    "negative": 0
})

print("\nFinal Dataset:")
print(data.head())

data.to_csv(
    r"D:\Jupyter\NLP Skills\Hometasks\cleaned_IMDB_reviews.csv",
    index=False
)

print("\nCleaned dataset saved successfully.")
```
